# Fase 1 : Engenharia de Features Avançada

**Objetivo:** O objetivo desta fase é enriquecer nossa base de dados, indo além dos números brutos sorteados. Nós transformamos cada sorteio em um conjunto de características mensuráveis e estatisticamente relevantes. Este processo, conhecido como "Engenharia de Features", é o que nos permite entender o "DNA" de um resultado da Lotofácil e, posteriormente, avaliar a qualidade de novas apostas.

**O que o código faz:** Para cada um dos mais de 3.400 sorteios do nosso histórico, o código calcula 6 "features" principais:
* **Soma Total:** A soma das 15 dezenas.
* **Qtd. Ímpares:** O equilíbrio entre números pares e ímpares.
* **Qtd. Repetidas:** Quantas dezenas vieram do sorteio imediatamente anterior.
* **Qtd. Moldura:** Quantas dezenas pertencem às bordas do volante.
* **Qtd. Primos:** A contagem de números primos no resultado.
* **Maior Sequência:** O maior número de dezenas consecutivas (ex: 07, 08, 09 = sequência de tamanho 3).

**Resultado Final:** Ao final da execução, temos o DataFrame `df_analise`. Esta é a nossa base de conhecimento completa, uma tabela que contém não apenas os resultados, mas também o perfil estatístico detalhado de cada um deles, pronta para ser usada na criação do nosso Sistema de Pontuação.

```python
# --- Bloco 1: Engenharia de Features Avançada ---

# Garante que estamos trabalhando com o dataframe limpo
df_analise = df_limpo.copy()
colunas_bolas = [f'bola_{i}' for i in range(1, 16)]

# 1. SOMA E PARES/ÍMPARES (Como antes)
df_analise['soma_total'] = df_analise[colunas_bolas].sum(axis=1)
df_analise['qtd_impares'] = df_analise[colunas_bolas].apply(lambda x: (x % 2 != 0).sum(), axis=1)

# 2. REPETIDAS E MOLDURA (Como antes)
# Repetidas
repetidas = []
for i in range(1, len(df_limpo)):
    sorteio_atual = set(df_limpo.iloc[i])
    sorteio_anterior = set(df_limpo.iloc[i-1])
    num_repetidas = len(sorteio_atual.intersection(sorteio_anterior))
    repetidas.append(num_repetidas)
df_analise['qtd_repetidas'] = [None] + repetidas
df_analise['qtd_repetidas'].fillna(int(df_analise['qtd_repetidas'].mean()), inplace=True)
df_analise['qtd_repetidas'] = df_analise['qtd_repetidas'].astype(int)

# Moldura
moldura = {1, 2, 3, 4, 5, 6, 10, 11, 15, 16, 20, 21, 22, 23, 24, 25}
df_analise['qtd_moldura'] = df_analise[colunas_bolas].apply(lambda x: len(set(x).intersection(moldura)), axis=1)

# 3. NOVAS FEATURES
# Números Primos
primos = {2, 3, 5, 7, 11, 13, 17, 19, 23}
df_analise['qtd_primos'] = df_analise[colunas_bolas].apply(lambda x: len(set(x).intersection(primos)), axis=1)

# Maior Sequência
def calcular_maior_sequencia(numeros):
    numeros_ordenados = sorted(list(numeros))
    maior_seq = 0
    seq_atual = 1
    if not numeros_ordenados:
        return 0
    for i in range(1, len(numeros_ordenados)):
        if numeros_ordenados[i] == numeros_ordenados[i-1] + 1:
            seq_atual += 1
        else:
            maior_seq = max(maior_seq, seq_atual)
            seq_atual = 1
    maior_seq = max(maior_seq, seq_atual)
    return maior_seq if maior_seq > 1 else 0

df_analise['maior_sequencia'] = df_analise[colunas_bolas].apply(calcular_maior_sequencia, axis=1)

# --- Verificando o DataFrame final ---
print("✅ DataFrame de Análise criado com todas as 6 features:")
display(df_analise.head())
```

# Fase 2: Definindo a Tabela de Pontuação

**Objetivo:** O objetivo desta fase é quantificar os padrões que observamos. Com as 6 features já calculadas para todo o histórico, precisamos agora entender a **distribuição** de cada uma delas. Qual é a soma mais comum? Qual a faixa principal de dezenas ímpares? Esta análise nos dará os parâmetros numéricos para construir um sistema de pontuação justo e baseado em dados.

**O que o código faz:** O comando `df_analise.describe()` funciona como um "raio-x estatístico" do nosso DataFrame. Ele calcula e exibe as métricas mais importantes para cada uma das 6 features, como:
* **`mean` (média)** e **`50%` (mediana):** Nos mostram o "centro da distribuição", ou seja, o valor mais comum e esperado para cada característica.
* **`25%`** e **`75%` (quartis):** Definem o "intervalo principal", a faixa de valores onde 50% de todos os resultados históricos se concentram. Esta é a nossa "zona de alta probabilidade".

**Resultado Final:** O resultado é uma tabela de estatísticas que serve como nosso **guia de referência**. É com base nesses números que nós criamos a **"Tabela de Pontos"** do nosso sistema, atribuindo a pontuação máxima (10 pontos) para os valores mais comuns (próximos da média/mediana) e pontuações menores para valores menos frequentes. Esta é a etapa que transforma a análise de dados brutos em regras concretas para o nosso modelo.

```python
# --- Bloco 2: Análise Estatística para o Sistema de Pontos ---

print("Estatísticas Descritivas para definir nosso sistema de pontos:")
display(df_analise.describe().loc[['mean', 'std', 'min', '25%', '50%', '75%', 'max']])
```

# Fase 3: Construindo o Sistema de Pontuação

**Objetivo:** O objetivo desta fase é traduzir a nossa "Tabela de Pontos", definida na fase anterior, em uma ferramenta funcional. Nós precisamos de um "motor" que possa pegar qualquer aposta, analisá-la e dar-lhe uma nota. Esta função será o núcleo lógico de todo o nosso projeto.

**O que o código faz:** O código define a função `calcular_pontuacao()`, que funciona como o cérebro do nosso sistema. Ela executa um processo de duas etapas:
1.  **Cálculo das Features:** Primeiro, para a aposta de 15 (ou 16) dezenas que ela recebe, a função calcula os valores de cada uma das 6 características (Soma, Ímpares, Repetidas, Moldura, Primos e Sequência).
2.  **Atribuição de Pontos:** Em seguida, ela usa uma série de condicionais (`if/elif`) que representam nossa Tabela de Pontos. Para cada característica calculada, ela atribui a pontuação correspondente (10 para "Excelente", 7 para "Bom", 4 para "Razoável" ou 0).

**Resultado Final:** O resultado da função é um único número: a **pontuação total (de 0 a 60)**, que representa a "Nota de Qualidade" da aposta. Esta função é a ferramenta de avaliação que será usada massivamente na próxima fase para analisar e classificar milhares de combinações, permitindo-nos, finalmente, encontrar os jogos mais promissores.

```python
# --- Bloco 3: Função de Pontuação (O Cérebro do Sistema) ---

def calcular_pontuacao(aposta, ultimo_sorteio):
    """
    Recebe uma aposta de 15 dezenas, calcula todas as suas features
    e retorna uma pontuação de 0 a 60 baseada na Tabela de Pontos.
    """
    # Etapa 1: Calcular todas as features da aposta fornecida
    soma = sum(aposta)
    impares = sum(1 for d in aposta if d % 2 != 0)
    primos_na_aposta = len(set(aposta).intersection(primos))
    moldura_na_aposta = len(set(aposta).intersection(moldura))
    repetidas_na_aposta = len(set(aposta).intersection(ultimo_sorteio))
    sequencia_na_aposta = calcular_maior_sequencia(aposta)

    # Etapa 2: Calcular os pontos para cada feature usando a Tabela
    pontos = 0

    # Pontos da Soma
    if 190 <= soma <= 200: pontos += 10
    elif 183 <= soma <= 189 or 201 <= soma <= 208: pontos += 7
    elif 175 <= soma <= 182 or 209 <= soma <= 215: pontos += 4

    # Pontos de Ímpares
    if impares == 8: pontos += 10
    elif impares in [7, 9]: pontos += 7
    elif impares in [6, 10]: pontos += 4

    # Pontos de Repetidas
    if repetidas_na_aposta == 9: pontos += 10
    elif repetidas_na_aposta in [8, 10]: pontos += 7
    elif repetidas_na_aposta in [7, 11]: pontos += 4

    # Pontos de Moldura
    if moldura_na_aposta == 10: pontos += 10
    elif moldura_na_aposta in [9, 11]: pontos += 7
    elif moldura_na_aposta == 8: pontos += 4

    # Pontos de Primos
    if primos_na_aposta == 5: pontos += 10
    elif primos_na_aposta in [4, 6]: pontos += 7
    elif primos_na_aposta in [3, 7]: pontos += 4

    # Pontos de Sequência
    if sequencia_na_aposta == 5: pontos += 10
    elif sequencia_na_aposta in [4, 6]: pontos += 7
    elif sequencia_na_aposta in [3, 7]: pontos += 4

    return pontos

print("✅ Função `calcular_pontuacao` criada e pronta para ser usada!")
```

Se isso acontecer, é um forte indicativo de que nossa "Tabela de Pontos" e a lógica da função estão bem calibradas e alinhadas com os padrões que os resultados reais costumam seguir. # Fase 4: O Gerador de Jogos Ranqueados

### **Fase 4: Geração do Ranking de Apostas**

**Objetivo:** Esta é a principal fase de aplicação prática do nosso projeto. O objetivo é usar o "cérebro" que construímos na fase anterior (a função `calcular_pontuacao`) para analisar em massa todas as combinações possíveis derivadas da sua estratégia de 11 números fixos. O resultado final será uma lista classificada (ranking) com as apostas mais promissoras.

**O que o código faz:** A função `gerar_jogos_ranqueados()` é o motor principal do nosso sistema. Ela executa um processo de três etapas:
1.  **Geração de Combinações:** A função pega seus 11 números fixos e, a partir dos 14 números restantes no volante, gera matematicamente todas as 1.001 combinações possíveis para completar os jogos de 15 dezenas.
2.  **Pontuação em Massa:** Em um loop, ela passa cada um desses 1.001 jogos pela nossa função `calcular_pontuacao`, atribuindo uma "Nota de Qualidade" (de 0 a 60) para cada um.
3.  **Classificação (Ranking):** Após analisar todos os jogos, a função os organiza em um ranking, do jogo com a maior pontuação até o de menor pontuação.

**Resultado Final:** O resultado é uma lista clara e ordenada com as melhores apostas possíveis baseadas na sua estratégia. Você recebe um "Top N" (por exemplo, Top 15 ou Top 30) com os jogos de elite, mostrando o **Rank**, a **Pontuação** e as **dezenas** de cada um. Esta lista é o principal produto do nosso sistema, transformando milhões de possibilidades e horas de análise em uma pequena e poderosa lista de apostas estratégicas.

```python
# --- Bloco 4: Função Final - Gerador de Jogos Ranqueados ---

from itertools import combinations

def gerar_jogos_ranqueados(numeros_fixos, ultimo_sorteio, top_n=30):
    """
    Gera e classifica todas as 1001 combinações possíveis a partir de 11 números fixos,
    retornando os melhores jogos com base em um sistema de pontuação.
    """
    if len(set(numeros_fixos)) != 11:
        print("❌ Erro: Você deve fornecer exatamente 11 números fixos e únicos.")
        return

    print("Iniciando a geração e análise de 1001 jogos... Isso pode levar um momento.")

    # Prepara os conjuntos de números
    numeros_fixos_set = set(numeros_fixos)
    universo_total = set(range(1, 26))
    numeros_restantes = list(universo_total - numeros_fixos_set)

    combinacoes_de_4 = combinations(numeros_restantes, 4)

    lista_de_jogos_com_pontos = []

    # Loop para gerar, analisar e pontuar cada jogo
    for combo in combinacoes_de_4:
        aposta_final = sorted(list(numeros_fixos_set) + list(combo))

        # Usa nossa função de pontuação para calcular a "Nota de Qualidade"
        pontuacao = calcular_pontuacao(aposta_final, ultimo_sorteio)

        lista_de_jogos_com_pontos.append({
            "jogo": aposta_final,
            "pontos": pontuacao
        })

    # Classifica a lista inteira para mostrar os melhores jogos no topo
    lista_ranqueada = sorted(lista_de_jogos_com_pontos, key=lambda x: x['pontos'], reverse=True)

    # Exibe os resultados
    print("\n" + "="*40)
    print("✅ ANÁLISE CONCLUÍDA - JOGOS RECOMENDADOS")
    print("="*40 + "\n")
    print(f"Exibindo os {top_n} melhores jogos encontrados (de 1001 analisados):\n")

    for i, item in enumerate(lista_ranqueada[:top_n]):
        print(f"  🏆 Rank {i+1:2d} | Pontuação: {item['pontos']}/60 | Jogo: {item['jogo']}")

    # Estatísticas adicionais
    pontuacao_maxima = lista_ranqueada[0]['pontos']
    pontuacao_media = sum(item['pontos'] for item in lista_ranqueada) / len(lista_ranqueada)
    print("\n" + "-"*40)
    print(f"Pontuação Máxima encontrada: {pontuacao_maxima}")
    print(f"Pontuação Média de todos os 1001 jogos: {pontuacao_media:.2f}")
    print("-"*40)

    return lista_ranqueada # Adiciona o retorno da lista ranqueada

print("✅ Função `gerar_jogos_ranqueados` criada com sucesso!")
```

# **Gerando as 15 Melhores Combinações**

## Criando e Analizando Combinações

```python
# --- Bloco Final e Definitivo: Gerar, Exibir Ranking Parcial e Verificar Histórico Completo ---

import pandas as pd
from itertools import combinations

# ===================================================================
# 1. DEFINIÇÃO DE TODAS AS FERRAMENTAS (O SISTEMA COMPLETO)
# ===================================================================

# --- Constantes e Tabela de Pontos ---
MOLDURA = {1, 2, 3, 4, 5, 6, 10, 11, 15, 16, 20, 21, 22, 23, 24, 25}
PRIMOS = {2, 3, 5, 7, 11, 13, 17, 19, 23}
TABELA_DE_PONTOS = {
    'soma': {10: range(190, 201), 7: list(range(183, 190)) + list(range(201, 209)), 4: list(range(175, 183)) + list(range(209, 216))},
    'impares': {10: [8], 7: [7, 9], 4: [6, 10]},
    'repetidas': {10: [9], 7: [8, 10], 4: [7, 11]},
    'moldura': {10: [10], 7: [9, 11], 4: [8]},
    'primos': {10: [5], 7: [4, 6], 4: [3, 7]},
    'sequencia': {10: [5], 7: [4, 6], 4: [3, 7]}
}

# --- Funções Auxiliares e de Pontuação ---
def calcular_maior_sequencia(numeros):
    numeros_ordenados = sorted(list(numeros))
    maior_seq, seq_atual = 0, 1
    if not numeros_ordenados: return 0
    for i in range(1, len(numeros_ordenados)):
        if numeros_ordenados[i] == numeros_ordenados[i-1] + 1: seq_atual += 1
        else:
            maior_seq = max(maior_seq, seq_atual)
            seq_atual = 1
    maior_seq = max(maior_seq, seq_atual)
    return maior_seq if maior_seq > 1 else 0

def get_pontos(valor, tabela_feature):
    for pontos, faixa in tabela_feature.items():
        if valor in faixa: return pontos
    return 0

def calcular_pontuacao(aposta, ultimo_sorteio):
    pontos = 0
    pontos += get_pontos(sum(aposta), TABELA_DE_PONTOS['soma'])
    pontos += get_pontos(sum(1 for d in aposta if d % 2 != 0), TABELA_DE_PONTOS['impares'])
    pontos += get_pontos(len(set(aposta).intersection(ultimo_sorteio)), TABELA_DE_PONTOS['repetidas'])
    pontos += get_pontos(len(set(aposta).intersection(MOLDURA)), TABELA_DE_PONTOS['moldura'])
    pontos += get_pontos(len(set(aposta).intersection(PRIMOS)), TABELA_DE_PONTOS['primos'])
    pontos += get_pontos(calcular_maior_sequencia(aposta), TABELA_DE_PONTOS['sequencia'])
    return pontos

# --- Função Geradora de Jogos (MODIFICADA) ---
def gerar_jogos_ranqueados(numeros_fixos, ultimo_sorteio, top_n=15):
    if len(set(numeros_fixos)) != 11: return None
    print(f"Analisando e classificando todas as 1001 combinações...")
    numeros_fixos_set = set(numeros_fixos)
    numeros_restantes = list(set(range(1, 26)) - numeros_fixos_set)
    combinacoes_de_4 = combinations(numeros_restantes, 4)
    lista_de_jogos_com_pontos = []
    for combo in combinacoes_de_4:
        aposta_final = sorted(list(numeros_fixos_set) + list(combo))
        pontuacao = calcular_pontuacao(aposta_final, ultimo_sorteio)
        lista_de_jogos_com_pontos.append({"jogo": aposta_final, "pontos": pontuacao})

    lista_ranqueada = sorted(lista_de_jogos_com_pontos, key=lambda x: x['pontos'], reverse=True)

    # Exibe apenas a quantidade de jogos que o usuário pediu (top_n)
    print("\n" + "="*55)
    print(f"✅ RANKING DOS {top_n} MELHORES JOGOS ENCONTRADOS")
    print("="*55 + "\n")
    for i, item in enumerate(lista_ranqueada[:top_n]):
        porcentagem = (item['pontos'] / 60) * 100
        print(f"  🏆 Rank {i+1:2d} | Qualidade: {porcentagem:.1f}% | Jogo: {item['jogo']}")

    # --- ALTERAÇÃO PRINCIPAL AQUI ---
    # A função agora retorna a LISTA COMPLETA de 1001 jogos para a verificação.
    return lista_ranqueada

# --- Função de Verificação Histórica ---
def verificar_jogos_em_todo_historico(lista_de_jogos_gerados, df_historico):
    print("\n" + "#"*60)
    print("--- VERIFICAÇÃO HISTÓRICA (Analisando a lista completa de 1001 jogos) ---")
    print("#"*60)
    historico_set = {tuple(sorted(row)) for index, row in df_historico.iterrows()}
    jogo_encontrado = False

    for i, jogo_dict in enumerate(lista_de_jogos_gerados):
        jogo_atual = tuple(sorted(jogo_dict['jogo']))
        if jogo_atual in historico_set:
            print("\n" + "="*50)
            print("🎉🎉🎉 CORRESPONDÊNCIA HISTÓRICA ENCONTRADA! 🎉🎉🎉")
            print("="*50)
            print(f"O seu Jogo de Rank {i+1} já foi sorteado na história da Lotofácil!")
            print(f"Jogo: {sorted(list(jogo_atual))}")
            jogo_encontrado = True

    if not jogo_encontrado:
        print("\nNenhum dos seus 1001 jogos corresponde a um resultado de 15 pontos em toda a história da nossa base de dados.")

# ===================================================================
# 2. EXECUÇÃO DO PROJETO
# ===================================================================

# --- SUAS CONFIGURAÇÕES ---
meus_numeros_fixos = [2, 4, 6, 10, 11, 13, 15, 16, 20, 24, 25]
quantidade_de_jogos_para_EXIBIR = 15 # <-- Defina quantos jogos quer VER no ranking

# --- LÓGICA PRINCIPAL ---
try:
    ultimo_sorteio_da_base = set(df_limpo.iloc[-1])

    # ETAPA 1: Gera os jogos. A função EXIBE o ranking parcial...
    jogos_gerados_completos = gerar_jogos_ranqueados(meus_numeros_fixos, ultimo_sorteio_da_base, top_n=quantidade_de_jogos_para_EXIBIR)

    # ETAPA 2: ...mas a verificação é feita na LISTA COMPLETA de 1001 jogos retornada.
    if jogos_gerados_completos:
        verificar_jogos_em_todo_historico(jogos_gerados_completos, df_limpo)

except NameError as e:
    print(f"\n❌ ERRO CRÍTICO: Uma função ou variável necessária não foi definida: {e}")
    print("Por favor, certifique-se de ter executado os Blocos 1 e 2 do código completo pelo menos uma vez nesta sessão.")
```

```python
# --- Bloco Final Adaptado: Análise Histórica Completa para 16 Números ---

import pandas as pd
from itertools import combinations
from collections import defaultdict

# ===================================================================
# 1. DEFINIÇÃO DAS FERRAMENTAS (ADAPTADAS)
# ===================================================================
# (Funções de pontuação permanecem as mesmas, mas a de geração será a nova)

MOLDURA = {1, 2, 3, 4, 5, 6, 10, 11, 15, 16, 20, 21, 22, 23, 24, 25}
PRIMOS = {2, 3, 5, 7, 11, 13, 17, 19, 23}
TABELA_DE_PONTOS = {
    'soma': {10: range(190, 201), 7: list(range(183, 190)) + list(range(201, 209)), 4: list(range(175, 183)) + list(range(209, 216))},
    'impares': {10: [8], 7: [7, 9], 4: [6, 10]}, 'repetidas': {10: [9], 7: [8, 10], 4: [7, 11]},
    'moldura': {10: [10], 7: [9, 11], 4: [8]}, 'primos': {10: [5], 7: [4, 6], 4: [3, 7]},
    'sequencia': {10: [5], 7: [4, 6], 4: [3, 7]}
}
def calcular_maior_sequencia(n):
    n_ord = sorted(list(n)); m, c = 0, 1
    if not n_ord: return 0
    for i in range(1, len(n_ord)):
        if n_ord[i] == n_ord[i-1] + 1: c += 1
        else: m = max(m, c); c = 1
    m = max(m, c)
    return m if m > 1 else 0
def get_pontos(v, t):
    for p, f in t.items():
        if v in f: return p
    return 0
def calcular_pontuacao(a, u):
    # A pontuação de um jogo de 16 dezenas é mais complexa, mas para simplificar nosso modelo,
    # vamos manter a mesma lógica de pontuação, avaliando as características do conjunto.
    p = 0; p += get_pontos(sum(a), TABELA_DE_PONTOS['soma']); p += get_pontos(sum(1 for d in a if d % 2 != 0), TABELA_DE_PONTOS['impares'])
    p += get_pontos(len(set(a).intersection(u)), TABELA_DE_PONTOS['repetidas']); p += get_pontos(len(set(a).intersection(MOLDURA)), TABELA_DE_PONTOS['moldura'])
    p += get_pontos(len(set(a).intersection(PRIMOS)), TABELA_DE_PONTOS['primos']); p += get_pontos(calcular_maior_sequencia(a), TABELA_DE_PONTOS['sequencia'])
    return p

# --- Função Geradora de Jogos (ADAPTADA PARA 16 NÚMEROS) ---
def gerar_jogos_ranqueados_16_numeros(fixos, ultimo, top_n=2002):
    if len(set(fixos)) != 11: return None
    fixos_set = set(fixos); restantes = list(set(range(1, 26)) - fixos_set)
    # ADAPTAÇÃO: combinations(..., 5) para gerar jogos de 16 números
    combos = combinations(restantes, 5); lista_jogos = []
    for c in combos:
        aposta = sorted(list(fixos_set) + list(c))
        pontuacao = calcular_pontuacao(aposta, ultimo)
        lista_jogos.append({"jogo": aposta, "pontos": pontuacao})
    return sorted(lista_jogos, key=lambda x: x['pontos'], reverse=True)

# --- Função de Análise Histórica (ADAPTADA PARA JOGOS DE 16) ---
def analise_historica_completa_16_numeros(lista_de_jogos_gerados, df_historico):
    print("\n" + "#"*50)
    print("--- ANÁLISE HISTÓRICA COMPLETA DE PRÊMIOS (16 NÚMEROS) ---")
    print("#"*50)
    print(f"Isso pode levar vários minutos... Comparando {len(lista_de_jogos_gerados)} jogos contra {len(df_historico)} sorteios.")

    premios_encontrados = []
    contagem_de_premios = defaultdict(int)

    for i, jogo_dict in enumerate(lista_de_jogos_gerados):
        rank = i + 1
        jogo_set = set(jogo_dict['jogo'])

        for indice_hist, linha_hist in df_historico.iterrows():
            resultado_historico_set = set(linha_hist)
            acertos = len(jogo_set.intersection(resultado_historico_set))

            # LÓGICA DE PREMIAÇÃO PARA 16 NÚMEROS:
            # Com 16 números, você ganha com 11, 12, 13, 14 ou 15 acertos, mas os prêmios são múltiplos.
            # Ex: Acertar 15 -> 1 prêmio de 15 e 15 prêmios de 14.
            # Para simplificar, vamos contar apenas o número de acertos diretos.
            if acertos >= 11:
                contagem_de_premios[acertos] += 1
                premios_encontrados.append({
                    "rank_do_jogo": rank, "acertos": acertos, "concurso": indice_hist + 1
                })

    print("\n" + "="*50)
    print("      RESUMO GERAL DE PRÊMIOS HISTÓRICOS (16 NÚMEROS)")
    print("="*50)
    total_premios = sum(contagem_de_premios.values())
    if total_premios > 0:
        print(f"Sua lista de {len(lista_de_jogos_gerados)} jogos teria conquistado um total de {total_premios} prêmios (11+ acertos) na história.\n")
        print("Placar de prêmios histórico (acertos diretos):\n")
        for acertos in sorted(contagem_de_premios.keys(), reverse=True):
            print(f"  ➡️ {acertos} acertos: {contagem_de_premios[acertos]} vez(es)")
        print("\n" + "="*50)
        print("      DETALHAMENTO DOS MELHORES RESULTADOS")
        print("="*50)
        for premio in sorted(premios_encontrados, key=lambda x: x['acertos'], reverse=True)[:10]:
            print(f"  🏆 Jogo de Rank {premio['rank_do_jogo']} acertou {premio['acertos']} pontos no concurso {premio['concurso']}")
    else:
        print("Nenhum prêmio (11-15 acertos) foi encontrado em todo o histórico.")
    print("\nAnálise concluída.")

# ===================================================================
# 2. EXECUÇÃO DO PROJETO (ADAPTADO)
# ===================================================================

# --- SUAS CONFIGURAÇÕES ---
meus_numeros_fixos = [2, 4, 6, 10, 11, 13, 15, 16, 20, 24, 25]

# --- LÓGICA PRINCIPAL ---
try:
    print("--- ETAPA 1: Gerando sua lista mestra de 2.002 jogos de 16 números... ---")
    ultimo_sorteio_da_base = set(df_limpo.iloc[-1])
    jogos_gerados_completos = gerar_jogos_ranqueados_16_numeros(meus_numeros_fixos, ultimo_sorteio_da_base)

    if jogos_gerados_completos:
        print("--- ETAPA 2: Iniciando a análise histórica completa... ---")
        analise_historica_completa_16_numeros(jogos_gerados_completos, df_limpo)

except NameError as e:
    print(f"\n❌ ERRO CRÍTICO: Uma função ou variável necessária não foi definida: {e}")
    print("Por favor, certifique-se de ter executado os blocos iniciais para carregar os dados ('df_limpo').")
```

### **1. A Descoberta do "DNA" de um Sorteio Vencedor**
O valor mais fundamental descoberto é que os resultados da Lotofácil não são aleatórios em sua *estrutura*. A grande maioria dos jogos vencedores segue padrões estatísticos claros. O projeto identificou e quantificou o "DNA" de um jogo de sucesso através de 6 características principais:
* Soma Total das dezenas
* Equilíbrio de números Ímpares e Pares
* Quantidade de Dezenas Repetidas
* Quantidade de Dezenas na Moldura
* Quantidade de Números Primos
* Tamanho da Maior Sequência

### **2. A Capacidade de Medir a "Qualidade" de uma Aposta**
O sistema evoluiu para criar um **Modelo de Pontuação (de 0 a 60)**. Essa "Nota de Qualidade" (expressa em porcentagem, como 90%) tornou-se o principal trunfo do modelo, permitindo diferenciar objetivamente uma aposta comum de uma aposta de elite com base em seu alinhamento com os padrões históricos.

A descoberta mais valiosa do projeto foi a **validação histórica**. ### **4. A Estratégia Inteligente Supera a Sorte Pura**
O valor final é a mudança de mentalidade. Foi compreendido que, embora a probabilidade de um bilhete único não possa ser alterada, é possível otimizar a seleção de apostas. O valor do sistema desenvolvido é garantir que cada aposta feita esteja sempre no seleto grupo de combinações de alta qualidade, maximizando o potencial de cada jogo e substituindo a sorte pura por uma **estratégia informada por dados**.
